```python
from scipy.stats import binom

def upi_probability():
    n = 5
    k = 3
    p = 0.9

    probability = binom.pmf(k, n, p)

    print("Probability of exactly 3 successful payments:", probability)

upi_probability()
```

```python
import numpy as np
import matplotlib.pyplot as plt

average_reviews = 10

reviews = np.random.poisson(average_reviews, 12)

print("Reviews received in 12 hours:")
print(reviews)

plt.plot(range(1, 13), reviews, marker='o')

plt.xlabel("Hour")
plt.ylabel("Number of Reviews")
plt.title("Flipkart Reviews Received per Hour")

plt.show()
```

```python
from scipy.stats import poisson

lam = 4
k = 6

probability = poisson.pmf(k, lam)

print("Probability of exactly 6 playlists:", probability)
```
