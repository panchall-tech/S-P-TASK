```python

sample_space = []

for die in range(1, 7):
    sample_space.append((die, "H"))
    sample_space.append((die, "T"))

print("Sample Space:")
print(sample_space)

print("Total outcomes:", len(sample_space))
```

```python
# Question 2: Probability of drawing an even number

total_cards = 10
even_cards = 5

probability = even_cards / total_cards

print("Number of even cards:", even_cards)
print("Total cards:", total_cards)
print("Probability of even number:", probability)
```

```python
# Question 3: Probability of NOT getting an Arijit Singh song

total_songs = 8
arijit_songs = 3

not_arijit_songs = total_songs - arijit_songs

probability = not_arijit_songs / total_songs

print("Songs not by Arijit Singh:", not_arijit_songs)
print("Total songs:", total_songs)
print("Probability:", probability)
```

```python
# Question 4: Probability of Pizza or Burger

pizza = 0.3
burger = 0.4
both = 0.1

probability = pizza + burger - both

print("Probability of Pizza or Burger:", probability)
```

```python
# Question 5: Conditional Probability

normal_probability = 0.2
mobile_interested_probability = 0.5

print("Normal probability of adding a phone:", normal_probability)
print("Probability when interested in mobiles:", mobile_interested_probability)

print("\nExplanation:")
print("Conditional probability means the probability changes")
print("when we have additional information.")
print("Here, the probability of adding a phone increases")
print("from 0.2 to 0.5 when we know the user is interested in mobiles.")
```
