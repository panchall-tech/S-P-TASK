```python
import numpy as np
from scipy.stats import f_oneway

# Dummy monthly order amounts for three customer segments
students = np.array([1200, 1500, 1300, 1600, 1400])
working_professionals = np.array([2500, 2800, 2700, 3000, 2600])
retirees = np.array([1800, 2000, 1900, 2100, 1950])

# Perform one-way ANOVA
f_statistic, p_value = f_oneway(
    students,
    working_professionals,
    retirees
)

# Display results
print("Zomato Monthly Order Amount ANOVA")
print("----------------------------------")

print("F-statistic:", f_statistic)
print("P-value:", p_value)

# Interpretation
if p_value < 0.05:
    print("\nConclusion:")
    print("There is a statistically significant difference")
    print("in average monthly order amounts among the three segments.")
else:
    print("\nConclusion:")
    print("There is no statistically significant difference")
    print("in average monthly order amounts among the three segments.")
```

```python
import pandas as pd
import numpy as np

# Read CSV file
df = pd.read_csv("purchase_data.csv")

# Calculate group means
group_means = df.groupby("age_group")["purchase_value"].mean()

# Calculate grand mean
grand_mean = df["purchase_value"].mean()

# Number of groups
k = len(group_means)

# Total number of observations
N = len(df)

# Number of observations in each group
group_counts = df.groupby("age_group")["purchase_value"].count()

# Between-group variance
between_variance = (
    sum(group_counts * (group_means - grand_mean) ** 2)
    / (k - 1)
)

# Within-group variance
within_variance = (
    df.groupby("age_group")["purchase_value"]
    .apply(lambda x: ((x - x.mean()) ** 2).sum())
    .sum()
    / (N - k)
)

print("Between-group variance:", between_variance)
print("Within-group variance:", within_variance)

print("\nExplanation:")
print("Between-group variance shows the differences between")
print("the average purchase values of the age groups.")

print("Within-group variance shows the variation of purchase")
print("values among customers within the same age group.")
```

```python
import pandas as pd
from scipy.stats import f_oneway

# Read CSV file
df = pd.read_csv("purchase_data.csv")

# Create groups
group1 = df[df["age_group"] == "18-25"]["purchase_value"]
group2 = df[df["age_group"] == "26-35"]["purchase_value"]
group3 = df[df["age_group"] == "36-50"]["purchase_value"]

# Perform ANOVA
f_statistic, p_value = f_oneway(group1, group2, group3)

print("F-statistic:", f_statistic)
print("P-value:", p_value)
```
