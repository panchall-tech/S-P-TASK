```python
import pandas as pd

cities = (['Ahmedabad'] * 500 +
          ['Surat'] * 300 +
          ['Vadodara'] * 200)

orders = pd.DataFrame({
    'Order_ID': range(1, 1001),
    'City': cities
})

sample_size = 60
sample_parts = []

for city, group in orders.groupby('City'):
    proportion = len(group) / len(orders)
    city_sample_size = round(proportion * sample_size)

    city_sample = group.sample(n=city_sample_size, random_state=42)
    sample_parts.append(city_sample)

stratified_sample = pd.concat(sample_parts)

print(stratified_sample)
print("\nSample count by city:")
print(stratified_sample['City'].value_counts())
```

```python
def systematic_sampling(reviews, interval=10):
    sample = []

    for i in range(interval - 1, len(reviews), interval):
        sample.append(reviews[i])

    return sample


reviews = [f"Review_{i}" for i in range(1, 501)]

sample = systematic_sampling(reviews, 10)

print("Systematic sample:")
print(sample)
```

```python
import random

fans = [
    ("Fan1", "CSK"), ("Fan2", "CSK"), ("Fan3", "CSK"),
    ("Fan4", "CSK"), ("Fan5", "CSK"), ("Fan6", "CSK"),
    ("Fan7", "MI"), ("Fan8", "MI"), ("Fan9", "MI"),
    ("Fan10", "MI"), ("Fan11", "MI"),
    ("Fan12", "RCB"), ("Fan13", "RCB"), ("Fan14", "RCB"),
    ("Fan15", "RCB"),
    ("Fan16", "GT"), ("Fan17", "GT"), ("Fan18", "GT"),
    ("Fan19", "KKR"), ("Fan20", "KKR")
]

sample_size = 40
groups = {}

for fan_id, team in fans:
    if team not in groups:
        groups[team] = []
    groups[team].append(fan_id)


total_fans = len(fans)

stratified_sample = []

for team, team_fans in groups.items():
    proportion = len(team_fans) / total_fans
    n = round(proportion * sample_size)

    selected = random.sample(team_fans, min(n, len(team_fans)))
    
    for fan in selected:
        stratified_sample.append((fan, team))

print("Stratified Sample:")
for fan in stratified_sample:
    print(fan)
```
