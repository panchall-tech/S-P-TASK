```python
from scipy import stats

delivery_times = [28, 32, 27, 35, 29, 31, 30, 33, 26, 34]

t_stat, p_value = stats.ttest_1samp(delivery_times, 30)

print("T-statistic:", t_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("The average delivery time is significantly different from 30 minutes.")
else:
    print("The average delivery time is not significantly different from 30 minutes.")
```

```python
from scipy import stats

before = [5000, 5200, 4800, 5500, 5100, 5300, 4900, 5000, 5150, 5250]

after = [6200, 6500, 6100, 6800, 6400, 6600, 6300, 6700, 6250, 6450]

t_stat, p_value = stats.ttest_ind(before, after)

print("T-statistic:", t_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("Steps increased significantly after using the fitness app.")
else:
    print("There is no significant increase in steps.")
```

```python
from scipy import stats

# Screen time in hours
before = [6.0, 5.5, 6.5, 7.0, 5.8, 6.2, 7.1, 6.8, 5.9, 6.4]

after = [5.0, 4.8, 5.6, 5.9, 4.9, 5.3, 6.0, 5.7, 5.1, 5.4]

t_stat, p_value = stats.ttest_rel(before, after)

print("T-statistic:", t_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("Focus Mode significantly reduced screen time.")
else:
    print("Focus Mode did not significantly reduce screen time.")
```

```python
from scipy import stats

zomato = [4.2, 4.0, 4.5, 3.8, 4.1, 4.3, 4.4, 3.9, 4.0, 4.6,
          4.2, 4.1, 3.7, 4.3, 4.5]

swiggy = [3.9, 4.1, 4.0, 3.8, 4.2, 4.0, 3.7, 4.1, 3.9, 4.3,
          4.0, 3.8, 4.2, 4.1, 3.9]

t_stat, p_value = stats.ttest_ind(zomato, swiggy)

print("T-statistic:", t_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("There is a significant difference between the average ratings.")
else:
    print("There is no significant difference between the average ratings.")
```

```python
from scipy import stats

transactions = [4, 5, 6, 3, 5, 7, 4, 6, 5, 4]

t_stat, p_value = stats.ttest_1samp(transactions, 5)

print("T-statistic:", t_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("The average number of transactions is significantly different from 5.")
else:
    print("The average number of transactions is not significantly different from 5.")
```
