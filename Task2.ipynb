```python
# Q4: Simple Measurement Scale Detection

from datetime import datetime


def detect_scale(data):
    # Empty data
    if len(data) == 0:
        return "Unknown"

    # Check for text values
    if all(isinstance(x, str) for x in data):

        # Ordered categories
        ordinal_values = [
            "low", "medium", "high",
            "poor", "average", "good", "excellent",
            "1 star", "2 star", "3 star", "4 star", "5 star"
        ]

        lower_data = [x.lower() for x in data]

        if all(x in ordinal_values for x in lower_data):
            return "Ordinal"

        # Check for date/time values
        try:
            for x in data:
                datetime.strptime(x, "%H:%M")
            return "Interval"
        except ValueError:
            pass

        # Other text values are nominal
        return "Nominal"

    # Check for numerical values
    if all(isinstance(x, (int, float)) for x in data):

        # Integer counts are usually ratio
        if all(isinstance(x, int) for x in data):
            return "Ratio"

        # Decimal measurements are usually ratio
        return "Ratio"

    return "Unknown"


# Examples from Flipkart order history

order_status = ["Pending", "Shipped", "Delivered"]
payment_method = ["UPI", "Credit Card", "Cash"]
product_quantity = [1, 2, 3, 4]
product_price = [499.99, 799.50, 1299.00]
departure_time = ["08:30", "12:45", "18:00"]

print("Order Status:", detect_scale(order_status))
print("Payment Method:", detect_scale(payment_method))
print("Product Quantity:", detect_scale(product_quantity))
print("Product Price:", detect_scale(product_price))
print("Departure Time:", detect_scale(departure_time))
```
