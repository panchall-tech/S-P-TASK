```python
import numpy as np

delivery_times = [25, 30, 35, 40, 28, 32, 45, 50, 27, 31,
                  38, 42, 29, 33, 36, 41, 48, 26, 34, 39]

q1 = np.percentile(delivery_times, 25)
q2 = np.percentile(delivery_times, 50)
q3 = np.percentile(delivery_times, 75)

print("25th Percentile (Q1):", q1)
print("50th Percentile (Q2):", q2)
print("75th Percentile (Q3):", q3)
```

```python
import numpy as np

steps = [6500, 7200, 8000, 6000, 9000, 7500, 8200, 7000,
         6800, 7600, 8500, 6200, 7800, 8100, 7300, 6900,
         9500, 6400, 7700, 8300, 7100, 7400, 8000, 6700,
         8600, 7900, 7200, 8800, 6100, 7500]

Q1 = np.percentile(steps, 25)
Q3 = np.percentile(steps, 75)

IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

```python
import numpy as np

def detect_outliers(amounts):
    Q1 = np.percentile(amounts, 25)
    Q3 = np.percentile(amounts, 75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = [x for x in amounts if x < lower_limit or x > upper_limit]

    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower Limit:", lower_limit)
    print("Upper Limit:", upper_limit)
    print("Outliers:", outliers)

amounts = [250, 300, 350, 400, 450, 500, 275, 325, 380, 420,
           460, 510, 290, 340, 390, 430, 480, 520, 2000]

detect_outliers(amounts)
```

```python
import matplotlib.pyplot as plt

screen_time = [28, 32, 30, 35, 31, 29, 33, 55]

# Create boxplot
plt.boxplot(
    screen_time,
    patch_artist=True,
    flierprops=dict(marker='o', markersize=8)
)

plt.ylabel("Screen Time (Hours)")
plt.xlabel("Weekly Screen Time")
plt.title("Weekly Screen Time for the Past 8 Weeks")

plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.show()
```
