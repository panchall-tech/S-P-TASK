```python
steps = [5000, 6500, 7200, 4800, 8000, 7500, 6000]

mean = sum(steps) / len(steps)

print("Mean daily steps:", mean)
```

```python
delivery_times = [32, 28, 29, 45, 30, 31, 60, 30, 29]

def find_median(data):
    data = sorted(data)
    n = len(data)
    
    if n % 2 == 1:
        return data[n // 2]
    else:
        return (data[n // 2 - 1] + data[n // 2]) / 2

median = find_median(delivery_times)

print("Median delivery time:", median, "minutes")
```

```python
transactions = [100, 120, 105, 110, 5000, 115, 108]

mean = sum(transactions) / len(transactions)

sorted_transactions = sorted(transactions)
median = sorted_transactions[len(sorted_transactions) // 2]

print("Mean:", mean)
print("Median:", median)
```
