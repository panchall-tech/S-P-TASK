```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(loc=500, scale=100, size=1000)

# Plot the distribution
plt.hist(data, bins=30, edgecolor="black")

plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Normal Distribution (Mean = 500, SD = 100)")

plt.show()
```

```python
X = 9500
mean = 8000
std = 1200

z = (X - mean) / std

print("Z-score:", z)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Generate x values
x = np.linspace(-4, 4, 1000)

# Calculate probability density
y = norm.pdf(x, 0, 1)

# Plot normal curve
plt.plot(x, y)

# Shade area between -1 and 1
x_fill = np.linspace(-1, 1, 500)
y_fill = norm.pdf(x_fill, 0, 1)

plt.fill_between(x_fill, y_fill)

plt.xlabel("Z-score")
plt.ylabel("Probability Density")
plt.title("Standard Normal Distribution")

plt.show()
```

```python
from scipy.stats import norm

mean = 30
std = 5
x = 25

probability = norm.cdf(x, loc=mean, scale=std)

print("Probability:", probability)
print("Percentage:", probability * 100, "%")
```
