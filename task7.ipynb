```python
#3
def bayes_churn_probability(prior_churn, prob_email_given_churn, prob_email):
    return (prob_email_given_churn * prior_churn) / prob_email



result = bayes_churn_probability(0.15, 0.6, 0.25)

print("Probability of churn:", result)
print("Percentage:", result * 100, "%")
```

```python
# Q4: P(Fraud | Alert)

prob_fraud = 0.02
prob_alert_given_fraud = 0.90
prob_alert_given_normal = 0.05

# P(Normal)
prob_normal = 1 - prob_fraud

# Total probability of getting an alert
prob_alert = (
    prob_alert_given_fraud * prob_fraud
    + prob_alert_given_normal * prob_normal
)

# Bayes' Theorem
prob_fraud_given_alert = (
    prob_alert_given_fraud * prob_fraud
) / prob_alert

print("P(Fraud | Alert) =", prob_fraud_given_alert)
print("Percentage =", prob_fraud_given_alert * 100, "%")
```
