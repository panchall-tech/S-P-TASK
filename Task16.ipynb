```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "price": [500, 1000, 1500, 2000, 2500, 3000, 3500, 4000, 4500, 5000],
    "user_rating": [3.8, 4.0, 4.2, 4.1, 4.5, 4.3, 4.6, 4.4, 4.7, 4.8],
    "reviews": [120, 250, 400, 550, 700, 900, 1100, 1350, 1600, 2000]
}

df = pd.DataFrame(data)

corr_matrix = df.corr()

# Create heatmap
sns.heatmap(
    corr_matrix,
    annot=True,
    cmap="coolwarm",
    cbar=True
)

plt.title("Flipkart Product Correlation Matrix")
plt.show()
```
