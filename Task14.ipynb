```python
import pandas as pd

data = {
    "Food Delivery App": [
        "Zomato", "Zomato", "Zomato", "Zomato", "Zomato",
        "Zomato", "Zomato", "Zomato", "Zomato",
        "Swiggy", "Swiggy", "Swiggy", "Swiggy", "Swiggy", "Swiggy", "Swiggy",
        "Uber Eats", "Uber Eats", "Uber Eats", "Uber Eats"
    ],
    
    "Payment Method": [
        "Paytm", "Paytm", "Paytm", "Paytm",
        "PhonePe", "PhonePe", "PhonePe",
        "Credit Card", "Credit Card",
        
        "Paytm", "Paytm",
        "PhonePe", "PhonePe", "PhonePe", "PhonePe",
        "Credit Card",
        
        "Paytm",
        "PhonePe", "PhonePe",
        "Credit Card"
    ]
}

df = pd.DataFrame(data)

contingency_table = pd.crosstab(
    df["Food Delivery App"],
    df["Payment Method"]
)

print("Contingency Table:")
print(contingency_table)
```

```python
import numpy as np

observed = np.array([
    [12, 8, 10],
    [18, 7, 15]
])

# Row totals
row_totals = observed.sum(axis=1)

# Column totals
column_totals = observed.sum(axis=0)

# Grand total
grand_total = observed.sum()

expected = np.outer(row_totals, column_totals) / grand_total

print("Row Totals:", row_totals)
print("Column Totals:", column_totals)
print("Grand Total:", grand_total)

print("\nExpected Counts:")
print(expected)
```

```python
import pandas as pd
from scipy.stats import chi2_contingency

table = pd.DataFrame(
    [
        [4, 3, 2],
        [2, 4, 1],
        [1, 2, 1]
    ],
    index=["Zomato", "Swiggy", "Uber Eats"],
    columns=["Paytm", "PhonePe", "Credit Card"]
)

print("Observed Contingency Table:")
print(table)

# Perform Chi-Square test
chi2, p_value, degrees_of_freedom, expected = chi2_contingency(table)

print("\nExpected Counts:")
print(expected)

print("\nChi-Square Statistic:", chi2)
print("Degrees of Freedom:", degrees_of_freedom)
print("P-value:", p_value)

# Decision
alpha = 0.05

if p_value < alpha:
    print("\nConclusion: Reject the null hypothesis.")
    print("There is a significant association between app choice and payment method.")
else:
    print("\nConclusion: Fail to reject the null hypothesis.")
    print("There is no significant association between app choice and payment method.")
```

```python
from scipy.stats import chisquare

observed = [18, 12]

expected = [30 * 0.70, 30 * 0.30]

print("Observed Counts:", observed)
print("Expected Counts:", expected)

chi_square = sum(
    (o - e) ** 2 / e
    for o, e in zip(observed, expected)
)

print("\nChi-Square Statistic:", chi_square)

from scipy.stats import chi2

degrees_of_freedom = 1
p_value = 1 - chi2.cdf(chi_square, degrees_of_freedom)

print("Degrees of Freedom:", degrees_of_freedom)
print("P-value:", p_value)

# Decision
alpha = 0.05

if p_value < alpha:
    print("\nConclusion: Reject the company's claim.")
    print("There is significant evidence that the preference is different from 70%.")
else:
    print("\nConclusion: Fail to reject the company's claim.")
    print("There is not enough evidence to reject the 70% claim.")
```
