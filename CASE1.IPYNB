```python
import pandas as pd

# Read the CSV file
df = pd.read_csv("OnlineRetail.csv")

# Calculate total sales
df["Sales"] = df["Quantity"] * df["UnitPrice"]

# Calculate total sales by country
country_sales = df.groupby("Country")["Sales"].sum()

# Sort from highest to lowest
country_sales = country_sales.sort_values(ascending=False)

print("Total Sales by Country:")
print(country_sales)

# Get the top 3 countries
top_3 = country_sales.head(3)

print("\nTop 3 Countries by Total Sales:")

for country, sales in top_3.items():
    print(country, "=", round(sales, 2))
```

```python
import pandas as pd

# Read the CSV file
df = pd.read_csv("OnlineRetail.csv")

# Calculate Order Value
df["OrderValue"] = df["Quantity"] * df["UnitPrice"]

# Calculate total order value for each InvoiceNo
order_values = df.groupby("InvoiceNo")["OrderValue"].sum()

# Calculate Average Order Value
average_order_value = order_values.mean()

print("Average Order Value =", round(average_order_value, 2))

# Find invoices with order value above ₹10,000
high_value_orders = order_values[order_values > 10000]

print("\nInvoices with Order Value above ₹10,000:")

if len(high_value_orders) == 0:
    print("No invoices found above ₹10,000.")
else:
    print(high_value_orders)
```

```python
import pandas as pd
from scipy.stats import ttest_ind

# Read the CSV file
df = pd.read_csv("OnlineRetail.csv")

# Calculate Order Value
df["OrderValue"] = df["Quantity"] * df["UnitPrice"]

# Select France and Germany data
france = df[df["Country"] == "France"]
germany = df[df["Country"] == "Germany"]

# Calculate total order value for each InvoiceNo
france_orders = france.groupby("InvoiceNo")["OrderValue"].sum()
germany_orders = germany.groupby("InvoiceNo")["OrderValue"].sum()

# Perform t-test assuming unequal variances
t_stat, p_value = ttest_ind(
    france_orders,
    germany_orders,
    equal_var=False
)

# Display results
print("France Average Order Value:", round(france_orders.mean(), 2))
print("Germany Average Order Value:", round(germany_orders.mean(), 2))

print("\nT-Statistic:", round(t_stat, 4))
print("P-Value:", round(p_value, 4))

# Check statistical significance
if p_value < 0.05:
    print("\nResult: The difference is statistically significant at the 0.05 level.")
else:
    print("\nResult: The difference is NOT statistically significant at the 0.05 level.")
```

```python
import pandas as pd

# Read the CSV file
df = pd.read_csv("OnlineRetail.csv")

# Calculate Order Value
df["OrderValue"] = df["Quantity"] * df["UnitPrice"]

# 1. Total sales by country
country_sales = df.groupby("Country")["OrderValue"].sum()
country_sales = country_sales.sort_values(ascending=False)

# 2. Average Order Value
order_values = df.groupby("InvoiceNo")["OrderValue"].sum()
average_order_value = order_values.mean()

# 3. Average quantity per transaction
average_quantity = df["Quantity"].mean()

# 4. Highest-value invoice
highest_invoice = order_values.idxmax()
highest_order_value = order_values.max()

# 5. Top 3 countries
top_3 = country_sales.head(3)

print("========== CUSTOMER PURCHASING INSIGHTS ==========\n")

print("1. Top 3 Countries by Total Sales:")
for country, sales in top_3.items():
    print(country, "=", round(sales, 2))

print("\n2. Average Order Value:")
print(round(average_order_value, 2))

print("\n3. Average Quantity per Transaction:")
print(round(average_quantity, 2))

print("\n4. Highest Value Invoice:")
print(highest_invoice, "=", round(highest_order_value, 2))

print("\n5. Overall Purchasing Pattern:")
print("Customers from the top countries contribute the highest sales.")
print("The average order value shows the typical amount spent per invoice.")
print("High-value invoices represent customers with larger purchases.")
```
