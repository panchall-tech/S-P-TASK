```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

population = np.random.exponential(scale=1000, size=10000)

sample_means = []

for i in range(1000):
    sample = np.random.choice(population, size=30, replace=True)
    sample_means.append(np.mean(sample))

# Plot the sampling distribution of means
plt.hist(sample_means, bins=30, edgecolor='black')

plt.xlabel("Sample Mean of Likes")
plt.ylabel("Frequency")
plt.title("Central Limit Theorem - Sample Size = 30")

plt.show()
```

```python
#4
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

population = np.random.exponential(scale=1000, size=10000)

# -----------------------------
# -----------------------------
sample_means_5 = []

for i in range(1000):
    sample = np.random.choice(population, size=5, replace=True)
    sample_means_5.append(np.mean(sample))

# -----------------------------
# Sample size = 30
# -----------------------------
sample_means_30 = []

for i in range(1000):
    sample = np.random.choice(population, size=30, replace=True)
    sample_means_30.append(np.mean(sample))

# Plot both distributions
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.hist(sample_means_5, bins=30, edgecolor='black')
plt.title("Sample Means (n = 5)")
plt.xlabel("Sample Mean")
plt.ylabel("Frequency")

plt.subplot(1, 2, 2)
plt.hist(sample_means_30, bins=30, edgecolor='black')
plt.title("Sample Means (n = 30)")
plt.xlabel("Sample Mean")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python
#5
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

population = np.random.exponential(scale=2, size=10000)

sample_means = []

for i in range(1000):
    sample = np.random.choice(population, size=30, replace=True)
    sample_means.append(np.mean(sample))

# Plot the sampling distribution
plt.hist(sample_means, bins=30, edgecolor='black')

plt.xlabel("Sample Mean")
plt.ylabel("Frequency")
plt.title("CLT Simulation using Exponential Distribution")

plt.show()
```

```python
#2The Central Limit Theorem (CLT) says that when we take many random samples from a population and calculate their means, the distribution of those means becomes approximately normal as the sample size increases.
```
